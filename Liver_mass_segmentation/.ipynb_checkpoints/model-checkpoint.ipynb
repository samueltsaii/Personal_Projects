```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
class Unet(nn.Module):
    def __init__(self, n_classes):
        super().__init__()
        # Encoder
        self.e11 = nn.Conv2d(1, 64, kernel_size = 3, padding = 1)
        self.e12 = nn.Conv2d(64, 64, kernel_size = 3, padding = 1)
        self.e1_pool = nn.MaxPool2d(kernel_size = 2, stride = 2)

        self.e21 = nn.Conv2d(64, 128, kernel_size = 3, padding = 1)
        self.e22 = nn.Conv2d(128, 128, kernel_size = 3, padding = 1)
        self.e2_pool = nn.MaxPool2d(kernel_size = 2, stride = 2)

        self.e31 = nn.Conv2d(128, 256, kernel_size = 3, padding = 1)
        self.e32 = nn.Conv2d(256, 256, kernel_size = 3, padding = 1)
        self.e3_pool = nn.MaxPool2d(kernel_size = 2, stride = 2)

        self.e41 = nn.Conv2d(256, 512, kernel_size = 3, padding = 1)
        self.e42 = nn.Conv2d(512, 512, kernel_size = 3, padding = 1)
        self.e3_pool = nn.MaxPool2d(kernel_size = 2, stride = 2)

        self.e51 = nn.Conv2d(512, 1024, kernel_size = 3, padding = 1)
        self.e52 = nn.Conv2d(1024, 1024, kernel_size = 3, padding = 1)
        
        # Decoder
        self.upconv1 = nn.ConvTranspose(1024, 512, kernel_size = 2)
        self.d11 = nn.Conv2d(1024, 512, kernel_size = 3, padding = 1)
        self.d12 = nn.Conv2d(512, 512, kernel_size = 3, padding = 1)

        self.upconv2 = nn.ConvTranspose(512, 256, kernel_size = 2)
        self.d21 = nn.Conv2d(512, 256, kernel_size = 3, padding = 1)
        self.d22 = nn.Conv2d(256, 256, kernel_size = 3, padding = 1)

        self.upconv3 = nn.ConvTranspose(256, 128, kernel_size = 2)
        self.d31 = nn.Conv2d(256, 128, kernel_size = 3, padding = 1)
        self.d32 = nn.Conv2d(128, 128, kernel_size = 3, padding = 1)

        self.upconv4 = nn.ConvTranspose(128, 64, kernel_size = 2)
        self.d41 = nn.Conv2d(128, 64, kernel_size = 3, padding = 1)
        self.d42 = nn.Conv2d(64, 64, kernel_size = 3, padding = 1)
        
        # Output 
        self.outconv = nn.Conv2d(64, n_classes, kernel_size = 1)

    def forward(self, x):
        # Encoder forward method
        x_e11 = relu(self.e11(x))
        x_e12 = relu(self.e12(x_e11))
        x_e1_pool = relu(self.e1_pool(x_e12))

        x_e21 = relu(self.e21(x_e1_pool))
        x_e22 = relu(self.e22(x_e21))
        x_e2_pool = relu(self.e2_pool(x_e22))

        x_e31 = relu(self.e31(x_e2_pool))
        x_e32 = relu(self.e32(x_e31))
        x_e3_pool = relu(self.e3_pool(x_e32))

        x_e41 = relu(self.e41(x_e3_pool))
        x_e42 = relu(self.e42(x_e41))
        x_e4_pool = relu(self.e2_pool(x_e42))

        x_e51 = relu(self.e51(x_e4_pool))
        x_e52 = relu(self.e52(x_e51))
        
        # Decoder forward method
        x_up1 = relu(self.conv1(x_e52))
        x_d11 = relu(self.d11(x_up1))
        x_d12 = relu(self.d21(x_d11))

        x_up2 = relu(self.conv2(x_d12))
        x_d21 = relu(self.d21(x_up2))
        x_d22 = relu(self.d22(x_d21))

        x_up3 = relu(self.conv3(x_d22))
        x_d31 = relu(self.d21(x_up3))
        x_d32 = relu(self.d31(x_d31))

        x_up4 = relu(self.conv4(x_d32))
        x_d41 = relu(self.d41(x_up4))
        x_d42 = relu(self.d42(x_d41))

        x_outconv = relu(self.outconv(x_d42))
```
